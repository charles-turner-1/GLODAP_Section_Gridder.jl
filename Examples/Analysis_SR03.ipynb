```julia
import Pkg
Pkg.activate("..")
Pkg.instantiate()
```

```julia
include("../src/GLODAP_Section_Gridder.jl")
using .GLODAP_Section_Gridder: grid_cruise
```

```julia
a05_1998_struct = grid_cruise(
    "33RO19980123",
    "A05",
    "G2theta",
    "isobaric"
)
```

```julia
horz_grid = a05_1998_struct.horz_grid
prs_grid = a05_1998_struct.vert_grid
gridded_data = a05_1998_struct.gridded_data

using Plots

Plots.heatmap(
    horz_grid,
    prs_grid,
    gridded_data;
    xlabel="Distance along section (km)",
    ylabel="Pressure (dbar)",
    title="Gridded Temperature: A05, 1998",
    yflip=true,
    colormap=:jet
)
```
